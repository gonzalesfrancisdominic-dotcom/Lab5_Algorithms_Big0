```python
import math

def get_student_inputs():
    """
    Replace these placeholders with your actual details.
    """
    # --- INPUT DATA HERE ---
    last_name = "Gonzales"
    student_id = "TUPM-26-4017"
    favorite_artist = "IV of Spades"
    # ----------------------------

    # 1. Calculate Student-Specific Values
    seed_digit = int(student_id[-1])
    id_sum = sum(int(digit) for digit in student_id if digit.isdigit())
    name_length = len(last_name)
    artist_length = len(favorite_artist)
    
    # Calculate ARTIST_SUM (A=1, B=2, ... Z=26)
    artist_sum = sum(ord(char.upper()) - ord('A') + 1 for char in favorite_artist if char.isalpha())

    print("--- Student Specific Inputs ---")
    print(f"Name: {last_name}")
    print(f"ID: {student_id}")
    print(f"Artist: {favorite_artist}")
    print(f"SEED_DIGIT: {seed_digit}")
    print(f"ID_SUM: {id_sum}")
    print(f"ARTIST_SUM: {artist_sum}")
    print("-" * 30)

    return seed_digit, id_sum, artist_sum


def generate_data(seed_digit, id_sum, artist_sum):
    """
    Generates the Original Dataset, Rotated Dataset, and Target Incident.
    """
    base_value = 100
    count = 20
    original_records = [base_value + (i * 10) for i in range(count)]
    
   
    rotation_point = seed_digit if seed_digit != 0 else 1
    rotation_point = rotation_point % len(original_records)

    rotated_records = original_records[rotation_point:] + original_records[:rotation_point]


    target = id_sum + artist_sum
    
    if target not in original_records:
        print(f"Note: Calculated target ({target}) not in base list. Adding it for test viability.")
        original_records.append(target)
        original_records.sort()
        rotated_records = original_records[rotation_point:] + original_records[:rotation_point]

    print(f"Generated Original Dataset: {original_records}")
    print(f"Rotation Point (Index): {rotation_point}")
    print(f"Generated Rotated Dataset:  {rotated_records}")
    print(f"Target Incident: {target}")
    print("-" * 30)
    
    return original_records, rotated_records, target

def binary_search_rotated(arr, target):
    """
    Search for target in rotated array.
    Returns: (index, comparisons)
    """
    low = 0
    high = len(arr) - 1
    comparisons = 0

    while low <= high:
        comparisons += 1
        mid = (low + high) // 2
        
        if arr[mid] == target:
            return mid, comparisons
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1

    return -1, comparisons

def linear_search(arr, target):
    """
    Standard Linear Search.
    Returns: (index, comparisons)
    """
    comparisons = 0
    for i in range(len(arr)):
        comparisons += 1
        if arr[i] == target:
            return i, comparisons
    return -1, comparisons

def main():
    seed, id_sum, art_sum = get_student_inputs()

    original, rotated, target = generate_data(seed, id_sum, art_sum)

    print("--- Binary Search (Optimized for Rotated Array) ---")
    idx_bin, comp_bin = binary_search_rotated(rotated, target)
    
    if idx_bin != -1:
        print(f"Target {target} found at index {idx_bin}.")
    else:
        print(f"Target {target} NOT found.")
    print(f"Comparisons made: {comp_bin}")
    print("-" * 30)

    print("--- Linear Search (Brute Force) ---")
    idx_lin, comp_lin = linear_search(rotated, target)
    
    if idx_lin != -1:
        print(f"Target {target} found at index {idx_lin}.")
    else:
        print(f"Target {target} NOT found.")
    print(f"Comparisons made: {comp_lin}")
    print("-" * 30)

    print("--- Analysis ---")
    print(f"Time Complexity:")
    print(f"  Binary Search: O(log n) - Much faster for large datasets.")
    print(f"  Linear Search: O(n)     - Slower, checks every element.")
    print(f"Space Complexity:")
    print(f"  Both Algorithms: O(1)   - Use constant extra space (iterative approach).")
    
    print("\nComparison Summary:")
    if comp_bin < comp_lin:
        print(f"Binary Search was more efficient (saved {comp_lin - comp_bin} comparisons).")
    elif comp_bin == comp_lin:
        print("Both algorithms took the same number of comparisons.")
    else:
        print("Linear Search was faster in this specific (small) instance.")

if __name__ == "__main__":
    main()
```

```python
def get_student_inputs():
    last_name = "Gonzales"
    student_id = "TUPM-26-4017"
    favorite_artist = "IV of Spades"

    seed_digit = int(student_id[-1])
    id_sum = sum(int(digit) for digit in student_id if digit.isdigit())
    name_length = len(last_name)
    artist_length = len(favorite_artist)

    return seed_digit, id_sum, artist_length, name_length

def generate_datasets(seed_digit, id_sum, artist_length, name_length):
    dataset_a = []
    val_a = seed_digit if seed_digit != 0 else 1
    for _ in range(id_sum):
        dataset_a.append(val_a)
        val_a += 2

    dataset_b = []
    val_b = artist_length if artist_length != 0 else 1
    for _ in range(name_length):
        dataset_b.append(val_b)
        val_b += 3

    return dataset_a, dataset_b

def consolidate_datasets(arr1, arr2):
    consolidated = []
    i = 0
    j = 0
    comparisons = 0

    while i < len(arr1) and j < len(arr2):
        comparisons += 1
        if arr1[i] <= arr2[j]:
            consolidated.append(arr1[i])
            i += 1
        else:
            consolidated.append(arr2[j])
            j += 1

    while i < len(arr1):
        consolidated.append(arr1[i])
        i += 1

    while j < len(arr2):
        consolidated.append(arr2[j])
        j += 1

    return consolidated, comparisons

def main():
    seed_digit, id_sum, artist_length, name_length = get_student_inputs()
    
    dataset_a, dataset_b = generate_datasets(seed_digit, id_sum, artist_length, name_length)
    
    consolidated, comparisons = consolidate_datasets(dataset_a, dataset_b)
    
    total_before = len(dataset_a) + len(dataset_b)
    total_after = len(consolidated)
    
    is_sorted = all(consolidated[i] <= consolidated[i+1] for i in range(len(consolidated)-1))

    print("Department A dataset:", dataset_a)
    print("Department B dataset:", dataset_b)
    print("Consolidated dataset:", consolidated)
    print("Number of records before consolidation:", total_before)
    print("Number of records after consolidation:", total_after)
    print("Comparison count:", comparisons)
    print("Ordering verification:", "Passed" if is_sorted else "Failed")
    print("Time complexity: O(N + M)")
    print("Space complexity: O(N + M)")

if __name__ == "__main__":
    main()
```

```python
import time
import random

def get_student_inputs():
    last_name = "Gonzales"
    student_id = "TUPM-26-4017"
    favorite_artist = "IV of Spades"

    seed_digit = int(student_id[-1])
    id_sum = sum(int(digit) for digit in student_id if digit.isdigit())
    name_length = len(last_name)
    artist_length = len(favorite_artist)
    artist_sum = sum(ord(char.upper()) - ord('A') + 1 for char in favorite_artist if char.isalpha())

    return seed_digit, id_sum, name_length, artist_length, artist_sum

def generate_orders(seed_digit, id_sum, name_length, artist_sum):
    size = (id_sum % 10) + 15 
    base_val = seed_digit + name_length + artist_sum
    
    random.seed(seed_digit + id_sum)
    orders = [random.randint(base_val, base_val + 100) for _ in range(size)]
    return orders

def bubble_sort(arr):
    a = arr.copy()
    n = len(a)
    comparisons = 0
    movements = 0
    
    start_time = time.perf_counter()
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                movements += 3
                swapped = True
        if not swapped:
            break
    end_time = time.perf_counter()
    return a, comparisons, movements, end_time - start_time

def selection_sort(arr):
    a = arr.copy()
    n = len(a)
    comparisons = 0
    movements = 0
    
    start_time = time.perf_counter()
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if a[j] < a[min_idx]:
                min_idx = j
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
            movements += 3
    end_time = time.perf_counter()
    return a, comparisons, movements, end_time - start_time

def insertion_sort(arr):
    a = arr.copy()
    n = len(a)
    comparisons = 0
    movements = 0
    
    start_time = time.perf_counter()
    for i in range(1, n):
        key = a[i]
        movements += 1
        j = i - 1
        while j >= 0:
            comparisons += 1
            if a[j] > key:
                a[j + 1] = a[j]
                movements += 1
                j -= 1
            else:
                break
        a[j + 1] = key
        movements += 1
    end_time = time.perf_counter()
    return a, comparisons, movements, end_time - start_time

def run_analysis():
    seed_digit, id_sum, name_length, artist_length, artist_sum = get_student_inputs()
    original_orders = generate_orders(seed_digit, id_sum, name_length, artist_sum)
    
    sorted_input = sorted(original_orders)
    reverse_input = sorted(original_orders, reverse=True)
    random_input = original_orders.copy()
    random.shuffle(random_input)

    conditions = {
        "Sorted": sorted_input,
        "Reverse": reverse_input,
        "Random": random_input
    }
    
    algorithms = {
        "Bubble Sort": bubble_sort,
        "Selection Sort": selection_sort,
        "Insertion Sort": insertion_sort
    }

    print("Generated order dataset:", original_orders)
    
    for cond_name, data in conditions.items():
        print(f"\n--- {cond_name} Input ---")
        print(f"Input Data: {data}")
        for algo_name, algo_func in algorithms.items():
            result, comps, movs, exec_time = algo_func(data)
            is_sorted = all(result[i] <= result[i+1] for i in range(len(result)-1))
            
            print(f"  {algo_name}:")
            print(f"    Comparisons: {comps}")
            print(f"    Movements:   {movs}")
            print(f"    Time (s):    {exec_time:.6f}")
            print(f"    Verified:    {is_sorted}")

    print("\n--- Big O Analysis & Observations ---")
    print("Bubble Sort:")
    print("  Best Case:    O(n)   - Array is already sorted (optimized with swapped flag).")
    print("  Average Case: O(n^2) - Elements are in random order.")
    print("  Worst Case:   O(n^2) - Array is reverse sorted (maximum swaps).")
    print("Selection Sort:")
    print("  Best/Avg/Worst: O(n^2) - Always performs the same number of comparisons regardless of initial order.")
    print("Insertion Sort:")
    print("  Best Case:    O(n)   - Array is already sorted (only 1 comparison per element).")
    print("  Average Case: O(n^2) - Elements are in random order.")
    print("  Worst Case:   O(n^2) - Array is reverse sorted (maximum shifts).")
    
    print("\nSpace Complexity: O(1) for all three algorithms (in-place sorting).")

if __name__ == "__main__":
    run_analysis()
```
